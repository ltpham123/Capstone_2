What is monthly total revenue for in-store sales in each of the two sales territories, over the full period covered by the data?

```python
# Monthly 
# # Northeast region (49)
# print(f"{assigned_region[0]} store IDs: {store_details[store_details["Region"] == assigned_region[0]]["Store Location"].unique()}")
# print(f"{assigned_region[0]} store count: {store_details[store_details["Region"] == assigned_region[0]]["Store Location"].count()}\n")

# # Northeast region (24)
# print(f"{assigned_region[1]} store IDs: {store_details[store_details["Region"] == assigned_region[1]]["Store Location"].unique()}")
# print(f"{assigned_region[1]} store count: {store_details[store_details["Region"] == assigned_region[1]]["Store Location"].count()}")

# monthly_sales = sales_storedetails_merge.groupby(sales_storedetails_merge["Transaction Date"]

# searched additional functions using AI, aggregate by month and year
sales_storedetails_merge["Transaction Date"] = pd.to_datetime(sales_storedetails_merge["Transaction Date"])
sales_storedetails_merge["Year_Month"] = sales_storedetails_merge["Transaction Date"].dt.to_period("M")

regional_sales = (
    sales_storedetails_merge[
        sales_storedetails_merge["Region"].isin(assigned_region)
        # sales_storedetails_merge["Region"]
    ]
    .groupby(["Region", "Year_Month"])["Sale Amount"]  # trying to aggregate by date, group by region then store location
    .sum()
    .reset_index(name="Revenue")
)

print(regional_sales)

# regional_sales.to_csv("regional_sales.csv", index=False)

# let's break it into quarters
# sales_storedetails_merge["Year_Quarter"] = sales_storedetails_merge["Transaction Date"].dt.to_period("Q")

# quarterly_sales = (
#     sales_storedetails_merge[
#         sales_storedetails_merge["Region"].isin(assigned_region)
#         # sales_storedetails_merge["Region"]
#     ]
#     .groupby(["Region", "Year_Quarter"])["Sale Amount"]  # trying to aggregate by date, group by region then store location
#     .sum()
#     .reset_index(name="Revenue")
# )

# print(quarterly_sales)
```

How would you rank the sales performance of each store in each sales territory? Which are the top-performing stores?

```python
# rank by revenue
store_revenue = (
    sales_storedetails_merge
    .groupby(["Region", "Store Location"])["Sale Amount"]  # by region then store location, can also substitute in for individual values
    .sum()
    .reset_index()
)

store_revenue["Sales_Rank"] = (
    store_revenue
    .groupby("Region")["Sale Amount"]
    .rank(method="dense", ascending=False)
)

store_revenue = store_revenue.sort_values(
    ["Region", "Sales_Rank", "Sale Amount"],
    ascending=[True, True, False]
)

# print(store_revenue)  # prints the ranking of all stores of their respective region

# top 5 stores per region
store_revenue_top5 = store_revenue[store_revenue["Sales_Rank"] <= 5]

print(store_revenue_top5)
```

Comparing the customer ID from the customer list data with the rewards ID from the sales data, who were the top customers in each sales territory?

```python
# left join sales with customer list and remove NaNs or missing IDs
# need to include region, so merge with store details too
# print(sales_storedetails_merge)

sales_storedetails_customer_merge = sales_storedetails_merge.merge(customer_list, left_on="RewardsID", right_on="cust_id", how="left")  # keep transaction list, remove any member who hasn't made a transaction
sales_storedetails_customer_merge = sales_storedetails_customer_merge.dropna(subset=["RewardsID"])  # drop nan
sales_storedetails_customer_merge["RewardsID"] = sales_storedetails_customer_merge["RewardsID"].astype(int)  # convert from float to integer

# print(sales_storedetails_customer_merge)

# Previously raised an error, checking if it exists:
# Error existed due to overwriting dataframe
# print("RewardsID" in store_sales.columns)
# print("cust_id" in customer_list.columns)  # cust_id, not cust_ID
# print(customer_list.columns)
# print(store_sales)

# print(sales_customer_merge.head())

# cleaning 3-merged-table table
# customer_revenue = customer_revenue.drop(columns=["Transaction Date", "Store ID", "Prod Num", "Year_Month", "cust_id", "date", "time", "email", "phone", "sms-opt-out "])
customer_revenue = sales_storedetails_customer_merge[["Region", "State", "Store Location", "RewardsID", "name", "Sale Amount"]]

# print(customer_revenue)

# borrowed from earlier for revenue
# rank by individual revenue
customer_revenue = (
    customer_revenue
    .groupby(["Region", "RewardsID", "name"])["Sale Amount"]  # by region then store location, can also substitute in for individual values
    .sum()
    .reset_index()
)

customer_revenue["Sales_Rank"] = (
    customer_revenue
    .groupby("Region")["Sale Amount"]
    .rank(method="dense", ascending=False)
)

customer_revenue = customer_revenue.sort_values(
    ["Region", "Sales_Rank", "RewardsID", "name", "Sale Amount"],
    ascending=[True, True, True, False, False]
)

# print(store_revenue)  # prints the ranking of all stores of their respective region

# top 5 stores per region
customer_revenue_top5 = customer_revenue[customer_revenue["Sales_Rank"] <= 5]

print(customer_revenue_top5)
```

What is the number of transactions per month by product category in each assigned territory? What is total sales revenue per month by category? What might this tell you about the most popular products, and where could there be opportunity for growth?

```python
# transaction count, aggregate similarly to region revenue
# join product details (category) to sales table
# print(store_sales.head(1))
# print(products.head(1))
# print(sales_storedetails_merge)

sales_products_merge = (
    sales_storedetails_merge
    .merge(products, on="Prod Num", how="left")
    .merge(product_categories, on="CategoryID", how="left")
)

# print(sales_products_merge)

# borrowed from monthly sales
# sales_products_merge["Transaction Date"] = pd.to_datetime(sales_products_merge["Transaction Date"])
# sales_products_merge["Year_Month"] = sales_products_merge["Transaction Date"].dt.to_period("M")

regional_inventory_summary = (
    sales_products_merge[
        sales_products_merge["Region"].isin(assigned_region)
    ]
    .groupby(["Region", "Year_Month", "Category",])  # by region then store location, can also substitute in for individual values
    .aggregate(  # multiple aggregation, doing both sum and count
        Revenue = ("Sale Amount", "sum"),
        Transaction_Count = ("Sale Amount", "size")
    )
    .reset_index()
)

# regional_inventory_summary["Revenue_Rank"] = (
#     regional_inventory_summary
#     .groupby("Region")["Revenue"]
#     .rank(method="dense", ascending=False)
# )

# regional_inventory_summary["Volume_Rank"] = (
#     regional_inventory_summary
#     .groupby("Region")["Transaction_Count"]
#     .rank(method="dense", ascending=False)
# )

# regional_inventory_summary = regional_inventory_summary.drop(columns=["Transaction Date", "Store ID", "RewardsID", "Region Director", "CategoryID", "SubcategoryID_x"])

# Northeast region
print(f"\n{assigned_region[0]}: ")
NE_monthly_revenue_top5 = (
    regional_inventory_summary[
        regional_inventory_summary["Region"] == assigned_region[0]
    ]
    .sort_values(["Region", "Year_Month", "Revenue"], ascending=[True, True, False])
    .groupby(["Region", "Year_Month"])
    .head(5)
)
print(NE_monthly_revenue_top5[["Region", "Year_Month", "Category", "Revenue"]].head(10))

NE_monthly_volume_top5 = (
    regional_inventory_summary[
        regional_inventory_summary["Region"] == assigned_region[0]
    ]
    .sort_values(["Region", "Year_Month", "Transaction_Count"], ascending=[True, True, False])
    .groupby(["Region", "Year_Month"])
    .head(5)
)

print(NE_monthly_volume_top5[["Region", "Year_Month", "Category", "Transaction_Count"]].head(10))

# South region
print(f"\n{assigned_region[1]}: ")
S_monthly_revenue_top5 = (
    regional_inventory_summary[
        regional_inventory_summary["Region"] == assigned_region[1]
    ]
    .sort_values(["Region", "Year_Month", "Revenue"], ascending=[True, True, False])
    .groupby(["Region", "Year_Month"])
    .head(5)
)
print(S_monthly_revenue_top5[["Region", "Year_Month", "Category", "Revenue"]].head(10))

S_monthly_volume_top5 = (
    regional_inventory_summary[
        regional_inventory_summary["Region"] == assigned_region[1]
    ]
    .sort_values(["Region", "Year_Month", "Transaction_Count"], ascending=[True, True, False])
    .groupby(["Region", "Year_Month"])
    .head(5)
)

print(S_monthly_volume_top5[["Region", "Year_Month", "Category", "Transaction_Count"]].head(10))
```
